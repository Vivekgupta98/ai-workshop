# Day 0 — One-State LangGraph with MLflow Gateway

Before running this notebook, complete the environment, MLflow, API-key, and gateway-endpoint setup in `README.md`.

## 1. Connect through MLflow Gateway

```python
from langchain_openai import ChatOpenAI

GATEWAY_URL = "http://127.0.0.1:5001/gateway/mlflow/v1"
GATEWAY_API_KEY = "not-needed"
GATEWAY_ENDPOINT = "workshop-gemini"

llm = ChatOpenAI(
    base_url=GATEWAY_URL,
    api_key=GATEWAY_API_KEY,
    model=GATEWAY_ENDPOINT,
)

reply = llm.invoke("Say: MLflow Gateway is working")
print(reply.content)
```

## 2. Build and print the graph

```python
from typing import TypedDict
from langgraph.graph import START, END, StateGraph

class State(TypedDict):
    user_input: str
    ai_reply: str

def call_ai(state: State):
    reply = llm.invoke(state["user_input"])
    return {"ai_reply": reply.content}

builder = StateGraph(State)
builder.add_node("call_ai", call_ai)
builder.add_edge(START, "call_ai")
builder.add_edge("call_ai", END)
graph = builder.compile()

print(graph.get_graph().draw_ascii())
```

## 3. Test the input and output

```python
user_input = input("You: ")
result = graph.invoke({"user_input": user_input})
print("AI:", result["ai_reply"])
```
